### Part 1.3 Web Scraping Weather Data
We will download and web scrape weather data.

1.  Run the following code.  Explain what `forecast_items` contains.

```python
#page = requests.get("http://forecast.weather.gov/MapClick.php?lat=37.7772&lon=-122.4168")
page = requests.get("https://forecast.weather.gov/MapClick.php?lat=41.3552&lon=-72.0996")
soup = BeautifulSoup(page.content, 'html.parser')
seven_day = soup.find(id="seven-day-forecast")
forecast_items = seven_day.find_all(class_="tombstone-container")
forecast_items
```

On the web page, these are the rectangular areas (shaped roughly like 'tombstones') that contain the day/time, weather condition, graphic, and temperature for tonight and the next four days.

Add `.get_text()` to the end of each command.  Print these three variables.  

```python
cur_tombstone = forecast_items[0]
period = cur_tombstone.find(class_="period-name").get_text()
short_desc = cur_tombstone.find(class_="short-desc").get_text()
temp = cur_tombstone.find(class_="temp").get_text()

print(period)
print(short_desc)
print(temp)
```

### Next we will extract all of the 7-day forecast info and store it in a dataframe.
Run any code shown below.

```python
period_tags = seven_day.select(".tombstone-container .period-name")
periods = [pt.get_text() for pt in period_tags]

short_descs = [sd.get_text() for sd in seven_day.select(".tombstone-container .short-desc")]
temps = [t.get_text() for t in seven_day.select(".tombstone-container .temp")]
descs = [d["title"] for d in seven_day.select(".tombstone-container img")]

print(periods,'\n')
print(short_descs,'\n')
print(temps,'\n')
print(descs,'\n')
```

```python
import pandas as pd
weather = pd.DataFrame({
        "period": periods, 
        "short_desc": short_descs, 
        "temp": temps, 
        "desc":descs
    })
weather
```

The next block of code extracts the numeric values from the temperature column.

```python
temp_nums = weather["temp"].str.extract("(?P<temp_num>\d+)", expand=False)
weather["temp_num"] = temp_nums.astype('int')
temp_nums
```
