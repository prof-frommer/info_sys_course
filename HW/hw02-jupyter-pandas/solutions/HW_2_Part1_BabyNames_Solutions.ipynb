```python
# Load the babynames data file into a pandas data frame called "bnDataFrame"
df = pd.read_csv('babynames.csv')
```

# 1.  Data set range

## a.  What is the first year of the data set?

```python
def getFirstYr(bnDataFrame):
    first_yr = min(bnDataFrame['Year'])
    return first_yr                    
print(getFirstYr(df))
```

## b.  What is the last year of the data set?

```python
def getLastYr(bnDataFrame):
    last_yr = max(bnDataFrame['Year'])
    return last_yr
print(getLastYr(df))
```

# 2. How many unique _boy_ names are there in this dataset?

```python
def numUniqBoyNames(bnDataFrame):
    filter_b = bnDataFrame['Gender'] == 'M'
    boy_rows = bnDataFrame[filter_b]
    uniq_boy_names = boy_rows['Name'].unique()
    num_uniq_boy_names = len(uniq_boy_names)
    return num_uniq_boy_names                # RETURN A NUMBER (most kinds of number data types should do)
print(numUniqBoyNames(df))
```

# 3. How many babies were named Taylor in this dataset?  

```python
def numTaylors(bnDataFrame):
    filter_taylor = bnDataFrame['Name'] == 'Taylor'
    taylor_rows = bnDataFrame[filter_taylor]
    total_taylors = sum(taylor_rows['Count'])
    return total_taylors                 # RETURN A NUMBER (most kinds of number data types should do)
print(numTaylors(df))
```

# 4. How many more boys were named Logan than girls in 1998?

```python
def logan98(bnDataFrame):
    filter_1 = (bnDataFrame['Name'] == 'Logan') & (bnDataFrame['Year'] == 1998)
    logan_98_rows = bnDataFrame[filter_1] 
    logan_98_boys = logan_98_rows[logan_98_rows['Gender'] == 'M']['Count'].iloc[0] 
    logan_98_girls = logan_98_rows[logan_98_rows['Gender'] == 'F']['Count'].iloc[0]
    return logan_98_boys - logan_98_girls     # RETURN A NUMBER (most kinds of number data types should do)
print(logan98(df))
```

# 5. Top three female names in 1998

## a. What were the top three female names in 1998?

```python
def top3Fs_98(bnDataFrame):
    filter_98 = bnDataFrame['Year']   == 1998
    filter_F  = bnDataFrame['Gender'] == 'F'
    rows_F_1998 = bnDataFrame[filter_98 & filter_F]
    sorted_rows_F_1998 = rows_F_1998.sort_values(by='Count',ascending=False)
    top_3 = list(sorted_rows_F_1998.iloc[:3]['Name'])
    return top_3  # RETURN A LIST
print(top3Fs_98(df))
```

## b. What were the counts of the top three female names in 1998? 

```python
def top3F_Counts_98(bnDataFrame):  
    filter_98 = bnDataFrame['Year']   == 1998
    filter_F  = bnDataFrame['Gender'] == 'F'
    rows_F_1998 = bnDataFrame[filter_98 & filter_F]
    sorted_rows_F_1998 = rows_F_1998.sort_values(by='Count',ascending=False)
    top_3 = list(sorted_rows_F_1998.iloc[:3]['Count'])
    return top_3  # RETURN A LIST
print(top3F_Counts_98(df))
```

# 6. Unisex names

## a. What are the unisex names in this dataset?

```python
def unisexNames(bnDataFrame):
    b_names = bnDataFrame[bnDataFrame['Gender']=='M']['Name'].unique()
    g_names = bnDataFrame[bnDataFrame['Gender']=='F']['Name'].unique()
    uni_names = set(g_names).intersection(set(b_names))
    return uni_names              # RETURN A SET
uni_names = unisexNames(bnDataFrame)
```

```python
uni_names_list = sorted(list(uni_names))
print(uni_names_list[:5],uni_names_list[-5:])
```
