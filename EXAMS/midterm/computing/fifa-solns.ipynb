b) Find the name, nationality, club, height, and weight of the player with the maximum value for "Potential".

c) Create a new column called "Mix" that is the average of the two columns "Weak Foot" and "Skill Moves". Display your data frame so that this new column is visible.

```python
df['Mix'] = (df['Weak Foot'] + df['Skill Moves']) / 2
df.head()
```

d)	The “Overall” field provides a player’s current ability rating where higher numbers are better. “Position” indicates where a player plays on the field, e.g. LM means left midfielder. Find the average (mean) value of “Overall” for each “Position” and report these in descending order of “Overall” value. Hint: use groupby.

```python
by_club = df.groupby('Position')
by_club['Overall'].mean().sort_values(ascending=False)
```
