### Write and run SQL queries that query database tables you just set up to accomplish the following:

4. ... return the name and team of each player

```python
sql_command = """SELECT Player_Name, Team FROM players;"""
pd.read_sql_query(sql_command, connection)
```

5. ... show the team, mascot, and noise level for matchdays with a noise level of 8 or higher. Sort from highest to lowest noise level.



```python
sql_command = """SELECT Team, Mascot, Noise_Level 
                 FROM matchday
                 WHERE Noise_Level >= 8
                 ORDER BY Noise_Level DESC
                 ;"""
pd.read_sql_query(sql_command, connection)
```

6. ... show each player, their position, their team, and the team's signature snack. Include records for which the team signature snack is not found in the database.

```python
sql_command = """SELECT players.Player_Name, players.Position, matchday.Team, Signature_Snack
                 FROM matchday
                 RIGHT JOIN players
                 ON matchday.Team = players.Team
                 ;"""
pd.read_sql_query(sql_command, connection)
```

or use a left join:

```python
sql_command = """SELECT players.Player_Name, players.Position, players.Team, Signature_Snack
                 FROM players
                 left JOIN matchday
                 ON matchday.Team = players.Team
                 ;"""
pd.read_sql_query(sql_command, connection)
```
