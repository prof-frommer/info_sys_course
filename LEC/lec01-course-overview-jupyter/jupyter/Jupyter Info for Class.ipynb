### There is a lot of very good information about Jupyter (& IPython) and much more in Jake VanderPlas' book *Python Data Science Handbook*.  Here is a sampling:

* IPython - which is the console (or "shell") we used in Spyder in CML - is a related to the broader *Jupyter* notebook structure (which includes notebooks for other programming languages such as R, Julia, etc.).  Note the file suffix of the Jupyter notebooks:  .ipynb, which stands for IPython notebook.  When VanderPlas talks about IPython we can think of Jupyter.  See also the other explanation below.
* His whole manuscript was written as an IPython notebook
* "IPython is about using Python effectively for interactive scientific and data-intensive computing."
* "The Jupyter notebook is a browser-based graphical interface to the IPython shell" and adds to it a "rich set of dynamic display capabilities" such as "formatted text, static and dynamic visualizations, mathematical equations, JavaScript widgets, and much more" and can be saved so that others can run them on their systems.
* Jupyter works by launching a local web server and running a Python process to execute code
* "When a technologically minded person is asked to help a friend, family member, or colleague with a computer problem, most of the time it's less a matter of knowing the answer as much as knowing how to quickly find an unknown answer.  In data science it's the same: searchable web resources such as online documentation, mailing-list threads, and Stack Overflow answers contain a wealth of information..."
* Key help functions: ? for documentation, ?? for source code, TAB for autocomplete

### Other notes
* Jupyter notebooks are similar to Mathematica notebooks in that they mix:
    * code
    * code output: numerical values, plots and other visualizations
    * formatted text explaining what the notebook author is doing, analyzing the results, etc.
* "As of IPython 4.0, the language-agnostic parts of the project: the notebook format, message protocol, qtconsole, notebook web application, etc. have moved to new projects under the name Jupyter. IPython itself is focused on interactive Python, part of which is providing a Python kernel for Jupyter."  from [ipython.org](https://ipython.org/#jupyter-and-the-future-of-ipython)
