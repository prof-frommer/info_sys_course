### 1.1 Using a **numpy array** of your data

### 1.2 Making data with a dict¶

Here's an example of making data from a dictionary and turning it into a pandas DataFrame

```python
# Make PFE data using a dictionary.
# The key is a string which will be the column headers in our data frame.
# The value is a list of strings which will give us our rows in the data frame
# Leave Code in***
pfe_data = {'first_name': ['Casey', 'John', 'Suzie', 'Joe'], 'last_name': ['Williams', 'Miller', 'Smith', 'Lee'], 'fall': [250, 275, 190, 205], 'spring': [270, 265, 200, 210]}
pfe_data
```

```python
# extract the keys for use as column names
colNames = pfe_data.keys()
colNames
```

```python
# Make a dataframe object and view it
# Recall that dicts are not stored in any particular order.  By using the columns parameter, we can
#    speficy the order in which to show the dataframe columns if need be.  Otherwise, without setting 
#    the columns parameter, pandas will still read the column names from the dict keys (in some order).
pfe_df = pd.DataFrame(pfe_data,columns=colNames)
pfe_df
```
