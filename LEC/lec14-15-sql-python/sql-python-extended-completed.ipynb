## Part 1

We will start by creating a database, `company.db` from scratch. The next block of code will pretty much be the same no matter what. 

```python
# import the sqlite3 library
import sqlite3

# create a connection object
connection = sqlite3.connect("company.db")

# create a cursor object
cursor = connection.cursor()
```

We now have an empty SQLite database file called company.db, which is saved in the same location as this Jupyter notebook file.  Naturally, our next step is to create some tables.

Recall from our earlier exercises, our SQL commands to create tables:

```python
# if necessary:  
try:
    cursor.execute('DROP TABLE employee')
except:
    pass

sql_command = """
CREATE TABLE employee ( 
staff_number INTEGER PRIMARY KEY, 
fname VARCHAR(20), 
lname VARCHAR(30), 
gender CHAR(1), 
joining DATE,
birth_date DATE);"""

# note:  the above can also be written as this:
# sql_command = "CREATE TABLE employee ( staff_number INTEGER PRIMARY KEY, fname VARCHAR(20), lname VARCHAR(30), gender CHAR(1), joining DATE,birth_date DATE);"
# more info at: https://stackoverflow.com/questions/19479853/why-do-we-need-to-use-3-quotes-while-executing-sql-query-from-python-cursor
```

We've saved the SQL command to create an `employee` table to the variable `sql_command`.  In order to execute this SQL statement, the following is needed.  Think of this as the same as clicking "Run" in Access.

```python
cursor.execute(sql_command)
```

Note:  once a table has been created, you cannot rerun the `CREATE TABLE` SQL command.  You will have to drop the table and then you can recreate it.  To drop the `employee` table: `cursor.execute("""DROP TABLE employee;""")`

Now that we have created a table, let's populate it.  Recall from earlier, we can INSERT INTO the `employee` table using the following SQL statement:

```python
# insert data into employee table
sql_command = """INSERT INTO employee (staff_number, fname, lname, gender, birth_date)
    VALUES (123, "Frank", "Schiller", "m", "1955-08-17");"""

# execute SQL command
cursor.execute(sql_command)
```

When inserting into a table, you must also commit changes.  Think of this as clicking "save" in Access

```python
# commit changes to the SQL database
connection.commit()
```

Did anything happen? You can look at the table by running a `SELECT *` query and then "fetching" the results:

```python
sql_command = """SELECT * from employee;"""
cursor.execute(sql_command)
```

```python
results = cursor.fetchall()
print(results)
```

I know what you're thinking -- this isn't realistic to enter in record by record.  Here's a way to do several records at once:

```python
staff_data = [ ("345", "William", "Shakespeare", "m", "1961-10-25"),
               ("678", "Wendy", "Wu", "f", "1955-08-17"),
               ("910", "Jane", "Wall", "f", "1989-03-14"),
               ("112", "Stephen", "Rodriguez", "m", "1971-2-25"),
               ("213", "Taylor", "Dahl", "f", "1953-3-02")]
               
```

We will make use of some Python f strings for text formatting.  Quick example below to give you the idea.

```python
price = 49
print(f'For only {price} dollars!')
```

Now back to the SQL work.

```python
for record in staff_data:
    print('Current record is: ',record)
    num, first, last, gender, birthdate = record
    sql_command = f"""INSERT INTO employee (staff_number, fname, lname, gender, birth_date)
    VALUES ("{num}", "{first}", "{last}", "{gender}", "{birthdate}");"""
    print('SQL command to add it to the table is:\n',sql_command,'\n')
    cursor.execute(sql_command)

connection.commit() 
```

```python
# A. What are the e-mail addresses of the male customers?
'''YOUR CODE HERE'''
sql1 = """
SELECT [Email Address]
FROM Customers
WHERE gender = 'M';
"""
pd.read_sql_query(sql1,connection)
```

```python
# B. How many monthly items cost less than $20?
'''YOUR CODE HERE'''
sql2 = """
SELECT Count([Monthly?]) AS [Under $20 Monthly Items]
FROM Items
WHERE Price < 20 and [Monthly?]=1
;
"""
df = pd.read_sql_query(sql2,connection)
df
```

```python
# C. List the customer first and last name alongside each order number.
'''YOUR CODE HERE'''
sql3 = """
SELECT [First Name],[Last Name],[Order ID]
FROM Customers
INNER JOIN Orders
ON Orders.[Customer ID] = Customers.[Customer ID]
;
"""
df = pd.read_sql_query(sql3,connection)
df
```

```python
""" 4. Write code to create two new tables in the database one containing the order line item rows in OrderLineItem.csv that are for monthly items, 
and another for those that are non-monthly items. Use a loop with f strings, then verify the success of the code by reading the
tables back in from the database file/connection."""

# Start by reading the orders and items tables into a data frame again
orderlineitems_df = pd.read_sql_query('SELECT * from OrderLineItem',connection)
items_df = pd.read_sql_query('SELECT * from Items',connection)
print(orderlineitems_df)
print(items_df)
```

```python
# Next set up queries to get the monthly items only or non-monthly items only
# And write to a table name Monthly or Non_Monthly accordingly
for monthly in [1,0]:
    sql_command = f"""SELECT [Order ID],Items.[Item ID], [Quantity],  [Item Name], [Description], [Price] 
                    FROM OrderLineItem
                    JOIN Items
                    on OrderLineItem.[Item ID] = Items.[Item ID]
                    WHERE Items.[Monthly?] = {monthly}; """
    df = pd.read_sql_query(sql_command, connection)
    table_name = 'Non_' * (1-monthly) + 'Monthly'
    df.to_sql(table_name, connection, if_exists='replace',index=False)
    print(df.head())
                
```

```python
# Run this command to display names of all the tables in your database
cursor.execute("""SELECT name
                from sqlite_master
                where type = 'table';""")
print(cursor.fetchall())
```

Now check to see if the data was written to the tables by reading it back in from the tables.  Loop through the monthly or not possibilities, creating sql queries via f strings to select all records from each table and display the results for each.  Each record should appear only once in each table (or dataframe).

```python
for monthly in [0,1]:  
    table_name = 'Non_' * (1-monthly) + 'Monthly'
    sql_command = f"""SELECT *
                FROM {table_name}
                ;"""
    df = pd.read_sql_query(sql_command, connection)
    print(table_name)
    print(df)
```
